```python
import os, re, json, numpy as np, tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
from tensorflow.keras import layers, Sequential
from pathlib import Path
```

```python
DATA_PATH = Path("shakespear.txt")
text = DATA_PATH.read_text(encoding='utf-8-sig')[:1000000]
```

```python
text[:10000]
```

```python
text = re.sub(r"\[.*?\]\([^)]*\)", ' ',text,flags=re.S)
```

```python
text[:1000]
```

```python
words = re.findall(r"[a-zA-Z']+",text.lower())[:80_000]
```

```python
words
```

```python
len(words)
```

```python
' '.join(words[:10])
```

```python
tokenizer = Tokenizer(oov_token='<DOV>')
```

```python
tokenizer.fit_on_texts([' '.join(words)])
```

```python
vocab_size = len(tokenizer.word_index)+1
```

```python
vocab_size
```

```python
ids = tokenizer.texts_to_sequences([' '.join(words)])[0]
```

```python
ids
```

```python
window, sequences = 6, []
```

```python
for i in range(window, len(ids)):
    sequences.append(ids[i-window:i + 1])
```

```python
sequences = np.array(sequences,dtype=np.int32)
```

```python
sequences
```

```python
X,y = sequences[:,:-1],sequences[:,-1]
```

```python
X
```

```python
y
```

```python
split = int(len(X)*0.8)
X_train,y_train = X[:split],y[:split]
X_test,y_test = X[split:],y[split:]
```

```python
vocab_size
```

```python
len(X_train)
```

```python

from tensorflow.keras import Sequential, layers

# Rebuild the model
model = Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Embedding(input_dim=vocab_size, output_dim=32),
    layers.SimpleRNN(48),
    layers.Dense(vocab_size, activation="softmax")
])

# Correct compile configuration
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# Check label range
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("Vocabulary size:", vocab_size)
print("Minimum label:", y_train.min())
print("Maximum label:", y_train.max())

# Train
history = model.fit(
    X_train,
    y_train,
    validation_split=0.2,
    epochs=5,
    batch_size=256
)
```

```python
index_to_word = {i:w for w,i in tokenizer.word_index.items()}
```

```python
index_to_word
```

```python

def predict_next_word(prompt, top_k=5):
    ids = tokenizer.texts_to_sequences([prompt])[0]
    ids = pad_sequences([ids], maxlen=window, padding='pre')

    probabilities = model.predict(ids, verbose=0)[0]

    best = np.argsort(probabilities)[-top_k:][::-1]

    return [
        (
            index_to_word.get(int(i), '<unknown>'),
            float(probabilities[i])
        )
        for i in best
    ]
```

```python

prompt = 'I love to machine”'

print('Prompt:', prompt)

for word, probability in predict_next_word(prompt):
    print(f'{word:15} {probability:.2%}')
```
